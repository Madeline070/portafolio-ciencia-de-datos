```python
import pandas as pd

fact_sales = pd.read_csv("FACT_SALES (1) (4).csv")
dim_product = pd.read_excel("DIM_PRODUCT (1) (4).xlsx")
dim_segment = pd.read_excel("DIM_SEGMENT (1) (3).xlsx")
dim_calendar = pd.read_excel("DIM_CALENDAR (2) (3).xlsx")
dim_category = pd.read_csv("DIM_CATEGORY (2) (1).csv")

print("Todo cargado correctamente")
```

```python
print("Duplicados:")

print("dim_category:", dim_category.duplicated().sum())
print("dim_product:", dim_product.duplicated().sum())
print("dim_segment:", dim_segment.duplicated().sum())
print("dim_calendar:", dim_calendar.duplicated().sum())
print("fact_sales:", fact_sales.duplicated().sum())
```

```python
dim_category.drop_duplicates(inplace=True)
dim_product.drop_duplicates(inplace=True)
dim_segment.drop_duplicates(inplace=True)
dim_calendar.drop_duplicates(inplace=True)
fact_sales.drop_duplicates(inplace=True)
```

```python
fact_sales['ITEM_CODE_LIMPIO'] = fact_sales['ITEM_CODE'].str.extract('(\d+)')
```

```python
# Solo si WEEK es fecha
# dim_calendar['WEEK'] = pd.to_datetime(dim_calendar['WEEK'])
```

```python
dim_product.columns = dim_product.columns.str.strip()
dim_category.columns = dim_category.columns.str.strip()

print(dim_product.columns)
print(dim_category.columns)
```

```python
dim_product['CATEGORY'] = pd.to_numeric(dim_product['CATEGORY'], errors='coerce')
dim_category['CATEGORY'] = pd.to_numeric(dim_category['CATEGORY'], errors='coerce')
```

```python
df_product_cat = dim_product.merge(
    dim_category,
    on='CATEGORY',
    how='left'
)

df_product_cat.head()
```

```python
df_sales_calendar = fact_sales.merge(dim_calendar, on='WEEK', how='left')
```

```python
df_final = df_sales_calendar.copy()

df_final.to_csv('dataset_consolidado.csv', index=False)

print("Archivo guardado correctamente")
```
