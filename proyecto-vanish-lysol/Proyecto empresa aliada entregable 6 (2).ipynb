```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

```

```python
fact_sales = pd.read_csv("FACT_SALES (1).csv")
dim_product = pd.read_excel("DIM_PRODUCT (1).xlsx")
dim_calendar = pd.read_excel("DIM_CALENDAR (2).xlsx")

print("FACT_SALES:", fact_sales.shape)
print("DIM_PRODUCT:", dim_product.shape)
print("DIM_CALENDAR:", dim_calendar.shape)
```

```python
display(dim_product[[
    "MANUFACTURER",
    "BRAND",
    "ITEM",
    "ITEM_DESCRIPTION"
]].head(10))
```

```python
print("Ejemplos de ITEM_CODE en FACT_SALES:")
print(fact_sales["ITEM_CODE"].head(10).tolist())
```

```python
dim_product["ITEM_CODE"] = (
    dim_product["ITEM_DESCRIPTION"]
    .str.extract(r"(\d{10,13})", expand=False)
    .str.lstrip("0")
)

print("Código creado correctamente.")
display(
    dim_product[
        ["ITEM_CODE", "BRAND", "ITEM", "ITEM_DESCRIPTION"]
    ].head()
)
```

```python
fact_sales["ITEM_CODE_BASE"] = (
    fact_sales["ITEM_CODE"]
    .astype(str)
    .str.extract(r"(\d{10,13})", expand=False)
)

print("Código de ventas preparado.")
display(
    fact_sales[
        ["ITEM_CODE", "ITEM_CODE_BASE"]
    ].head(10)
)
```

```python
ventas_productos = fact_sales.merge(
    dim_product,
    left_on="ITEM_CODE_BASE",
    right_on="ITEM_CODE",
    how="left"
)

print("Dimensiones:", ventas_productos.shape)

display(
    ventas_productos[
        [
            "ITEM_CODE_x",
            "BRAND",
            "ITEM",
            "TOTAL_UNIT_SALES",
            "TOTAL_VALUE_SALES"
        ]
    ].head()
)
```

```python
ventas_clave = ventas_productos[
    ventas_productos["BRAND"].str.contains(
        "VANISH|LYSOL",
        case=False,
        na=False
    )
].copy()

print("Marcas encontradas:")
print(ventas_clave["BRAND"].unique())

print("\nNúmero de registros:")
print(len(ventas_clave))
```

```python
ventas_semanales = (
    ventas_clave
    .groupby(["WEEK", "BRAND"], as_index=False)
    ["TOTAL_UNIT_SALES"]
    .sum()
)

print("Dimensiones:", ventas_semanales.shape)

display(ventas_semanales.head(10))
```

```python
print("Semanas disponibles:")
print(ventas_semanales["WEEK"].nunique())

print("\nMarcas:")
print(ventas_semanales["BRAND"].unique())

print("\nRango de semanas:")
print(
    ventas_semanales["WEEK"].min(),
    "a",
    ventas_semanales["WEEK"].max()
)
```

```python
plt.figure(figsize=(12, 6))

for marca in ["VANISH", "LYSOL"]:
    datos = ventas_semanales[
        ventas_semanales["BRAND"] == marca
    ]
    
    plt.plot(
        datos["WEEK"],
        datos["TOTAL_UNIT_SALES"],
        label=marca
    )

plt.title("Ventas semanales de Vanish y Lysol")
plt.xlabel("Semana")
plt.ylabel("Unidades vendidas")
plt.legend()
plt.grid(axis="y", alpha=0.25)
plt.tight_layout()
plt.show()
```

```python
ventas_semanales["AÑO"] = (
    ventas_semanales["WEEK"]
    .str[-2:]
    .astype(int)
)

ventas_semanales["NUM_SEMANA"] = (
    ventas_semanales["WEEK"]
    .str[:2]
    .astype(int)
)

ventas_semanales = ventas_semanales.sort_values(
    ["AÑO", "NUM_SEMANA"]
).reset_index(drop=True)

print("Primeras semanas:")
print(ventas_semanales["WEEK"].unique()[:10])

print("\nÚltimas semanas:")
print(ventas_semanales["WEEK"].unique()[-10:])
```

```python
vanish = (
    ventas_semanales[
        ventas_semanales["BRAND"] == "VANISH"
    ][["WEEK", "TOTAL_UNIT_SALES"]]
    .reset_index(drop=True)
)

lysol = (
    ventas_semanales[
        ventas_semanales["BRAND"] == "LYSOL"
    ][["WEEK", "TOTAL_UNIT_SALES"]]
    .reset_index(drop=True)
)

print("Vanish:", len(vanish), "semanas")
print("Lysol:", len(lysol), "semanas")
```

```python
fig, axes = plt.subplots(2, 1, figsize=(12, 8))

axes[0].plot(vanish["TOTAL_UNIT_SALES"])
axes[0].set_title("Ventas semanales de Vanish")
axes[0].set_ylabel("Unidades vendidas")
axes[0].grid(axis="y", alpha=0.25)

axes[1].plot(lysol["TOTAL_UNIT_SALES"])
axes[1].set_title("Ventas semanales de Lysol")
axes[1].set_xlabel("Semana")
axes[1].set_ylabel("Unidades vendidas")
axes[1].grid(axis="y", alpha=0.25)

plt.tight_layout()
plt.show()
```

```python
train_size = 64

vanish_train = vanish["TOTAL_UNIT_SALES"].iloc[:train_size]
vanish_test = vanish["TOTAL_UNIT_SALES"].iloc[train_size:]

lysol_train = lysol["TOTAL_UNIT_SALES"].iloc[:train_size]
lysol_test = lysol["TOTAL_UNIT_SALES"].iloc[train_size:]

print("Vanish:")
print("Entrenamiento:", len(vanish_train))
print("Prueba:", len(vanish_test))

print("\nLysol:")
print("Entrenamiento:", len(lysol_train))
print("Prueba:", len(lysol_test))
```

```python
fig, axes = plt.subplots(2, 1, figsize=(12, 8))

axes[0].plot(range(64), vanish_train, label="Entrenamiento")
axes[0].plot(range(64, 80), vanish_test, label="Prueba")
axes[0].set_title("Vanish: entrenamiento y prueba")
axes[0].set_ylabel("Unidades vendidas")
axes[0].legend()
axes[0].grid(axis="y", alpha=0.25)

axes[1].plot(range(64), lysol_train, label="Entrenamiento")
axes[1].plot(range(64, 80), lysol_test, label="Prueba")
axes[1].set_title("Lysol: entrenamiento y prueba")
axes[1].set_xlabel("Semana")
axes[1].set_ylabel("Unidades vendidas")
axes[1].legend()
axes[1].grid(axis="y", alpha=0.25)

plt.tight_layout()
plt.show()
```

```python
from statsmodels.tsa.arima.model import ARIMA

print("ARIMA disponible correctamente.")
```

```python
def buscar_arima(serie):
    resultados = []

    for p in range(4):
        for d in range(2):
            for q in range(4):
                try:
                    modelo = ARIMA(
                        serie,
                        order=(p, d, q)
                    ).fit()

                    resultados.append({
                        "p": p,
                        "d": d,
                        "q": q,
                        "AIC": modelo.aic,
                        "BIC": modelo.bic
                    })

                except:
                    pass

    return pd.DataFrame(resultados).sort_values("AIC")


resultados_vanish = buscar_arima(vanish_train)
resultados_lysol = buscar_arima(lysol_train)

print("Mejores modelos para Vanish:")
display(resultados_vanish.head())

print("\nMejores modelos para Lysol:")
display(resultados_lysol.head())
```

```python
mejor_vanish = resultados_vanish.iloc[0]
mejor_lysol = resultados_lysol.iloc[0]

print("Mejor ARIMA para Vanish:")
print(
    f"ARIMA({int(mejor_vanish['p'])}, "
    f"{int(mejor_vanish['d'])}, "
    f"{int(mejor_vanish['q'])})"
)

print("\nMejor ARIMA para Lysol:")
print(
    f"ARIMA({int(mejor_lysol['p'])}, "
    f"{int(mejor_lysol['d'])}, "
    f"{int(mejor_lysol['q'])})"
)
```

```python
modelo_vanish = ARIMA(
    vanish_train,
    order=(
        int(mejor_vanish["p"]),
        int(mejor_vanish["d"]),
        int(mejor_vanish["q"])
    )
).fit()

modelo_lysol = ARIMA(
    lysol_train,
    order=(
        int(mejor_lysol["p"]),
        int(mejor_lysol["d"]),
        int(mejor_lysol["q"])
    )
).fit()

```

```python
pred_vanish = modelo_vanish.forecast(steps=len(vanish_test))
pred_lysol = modelo_lysol.forecast(steps=len(lysol_test))

print("Predicciones Vanish:")
display(pred_vanish)

print("\nPredicciones Lysol:")
display(pred_lysol)
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error

def calcular_metricas(real, pred):
    mse = mean_squared_error(real, pred)
    mae = mean_absolute_error(real, pred)
    mape = np.mean(
        np.abs((real - pred) / real)
    ) * 100
    
    return mse, mae, mape


mse_vanish, mae_vanish, mape_vanish = calcular_metricas(
    vanish_test,
    pred_vanish
)

mse_lysol, mae_lysol, mape_lysol = calcular_metricas(
    lysol_test,
    pred_lysol
)

metricas = pd.DataFrame({
    "Marca": ["VANISH", "LYSOL"],
    "MSE": [mse_vanish, mse_lysol],
    "MAE": [mae_vanish, mae_lysol],
    "MAPE (%)": [mape_vanish, mape_lysol]
})

display(metricas)
```

```python
fig, axes = plt.subplots(2, 1, figsize=(12, 8))

axes[0].plot(
    vanish_test.values,
    label="Ventas reales"
)
axes[0].plot(
    pred_vanish.values,
    label="Predicciones"
)
axes[0].set_title("Vanish: ventas reales vs. predichas")
axes[0].set_ylabel("Unidades vendidas")
axes[0].legend()
axes[0].grid(axis="y", alpha=0.25)

axes[1].plot(
    lysol_test.values,
    label="Ventas reales"
)
axes[1].plot(
    pred_lysol.values,
    label="Predicciones"
)
axes[1].set_title("Lysol: ventas reales vs. predichas")
axes[1].set_xlabel("Semana de prueba")
axes[1].set_ylabel("Unidades vendidas")
axes[1].legend()
axes[1].grid(axis="y", alpha=0.25)

plt.tight_layout()
plt.show()
```

```python
comparacion_modelos = pd.DataFrame({
    "Marca": ["VANISH", "LYSOL"],
    "Modelo": [
        f"ARIMA({int(mejor_vanish['p'])},{int(mejor_vanish['d'])},{int(mejor_vanish['q'])})",
        f"ARIMA({int(mejor_lysol['p'])},{int(mejor_lysol['d'])},{int(mejor_lysol['q'])})"
    ],
    "AIC": [
        modelo_vanish.aic,
        modelo_lysol.aic
    ],
    "BIC": [
        modelo_vanish.bic,
        modelo_lysol.bic
    ]
})

display(comparacion_modelos)
```

```python
mejor_bic_vanish = resultados_vanish.sort_values("BIC").iloc[0]
mejor_bic_lysol = resultados_lysol.sort_values("BIC").iloc[0]

print("Mejor modelo para VANISH según BIC:")
print(
    f"ARIMA({int(mejor_bic_vanish['p'])},"
    f"{int(mejor_bic_vanish['d'])},"
    f"{int(mejor_bic_vanish['q'])})"
)

print("\nMejor modelo para LYSOL según BIC:")
print(
    f"ARIMA({int(mejor_bic_lysol['p'])},"
    f"{int(mejor_bic_lysol['d'])},"
    f"{int(mejor_bic_lysol['q'])})"
)
```

```python
comparacion_criterios = pd.DataFrame({
    "Marca": ["VANISH", "LYSOL"],
    "Modelo_AIC": [
        f"ARIMA({int(mejor_vanish['p'])},{int(mejor_vanish['d'])},{int(mejor_vanish['q'])})",
        f"ARIMA({int(mejor_lysol['p'])},{int(mejor_lysol['d'])},{int(mejor_lysol['q'])})"
    ],
    "Modelo_BIC": [
        f"ARIMA({int(mejor_bic_vanish['p'])},{int(mejor_bic_vanish['d'])},{int(mejor_bic_vanish['q'])})",
        f"ARIMA({int(mejor_bic_lysol['p'])},{int(mejor_bic_lysol['d'])},{int(mejor_bic_lysol['q'])})"
    ]
})

display(comparacion_criterios)
```

```python
modelo_final_vanish = ARIMA(
    vanish["TOTAL_UNIT_SALES"],
    order=(
        int(mejor_vanish["p"]),
        int(mejor_vanish["d"]),
        int(mejor_vanish["q"])
    )
).fit()

modelo_final_lysol = ARIMA(
    lysol["TOTAL_UNIT_SALES"],
    order=(
        int(mejor_lysol["p"]),
        int(mejor_lysol["d"]),
        int(mejor_lysol["q"])
    )
).fit()

print("Modelos finales entrenados con las 80 semanas.")
```

```python
pasos_futuros = 12

pronostico_vanish = modelo_final_vanish.forecast(
    steps=pasos_futuros
)

pronostico_lysol = modelo_final_lysol.forecast(
    steps=pasos_futuros
)

print("Pronóstico futuro de VANISH:")
display(pronostico_vanish)

print("\nPronóstico futuro de LYSOL:")
display(pronostico_lysol)
```

```python
predicciones_futuras = pd.DataFrame({
    "Semana_futura": range(1, pasos_futuros + 1),
    "VANISH": pronostico_vanish.values,
    "LYSOL": pronostico_lysol.values
})

display(predicciones_futuras)
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    range(1, 13),
    pronostico_vanish.values,
    marker="o",
    label="VANISH"
)

plt.plot(
    range(1, 13),
    pronostico_lysol.values,
    marker="o",
    label="LYSOL"
)

plt.title("Predicción de ventas para las próximas 12 semanas")
plt.xlabel("Semana futura")
plt.ylabel("Unidades vendidas")
plt.legend()
plt.grid(axis="y", alpha=0.25)
plt.tight_layout()
plt.show()
```

## Resultados del modelo

Para realizar el pronóstico de ventas se utilizaron modelos ARIMA para las marcas VANISH y LYSOL.

Los datos se dividieron en 64 semanas para entrenamiento y 16 semanas para prueba.

El modelo seleccionado para VANISH fue ARIMA(3,1,0), mientras que para LYSOL fue ARIMA(0,1,0). La selección se realizó mediante la comparación de los criterios AIC y BIC.

## Validación y ajuste de los modelos

Los modelos fueron evaluados utilizando MSE, MAE y MAPE.

Para VANISH se obtuvo un MAE de 80.10 unidades y un MAPE de 11.87%.

Para LYSOL se obtuvo un MAE de 0.47 unidades y un MAPE de 10.23%.

La comparación mediante AIC y BIC confirmó los mismos modelos seleccionados para ambas marcas, por lo que no fue necesario realizar un cambio adicional en los parámetros.

## Pronóstico y recomendaciones

El pronóstico de las siguientes 12 semanas muestra un comportamiento relativamente estable.

Para VANISH se esperan ventas aproximadas entre 741 y 778 unidades por semana. Para LYSOL, el modelo proyecta aproximadamente 5.14 unidades por semana.

Estos resultados pueden utilizarse como referencia para la planeación de inventario y seguimiento de las ventas. Se recomienda comparar periódicamente las ventas reales con las predicciones y actualizar los modelos cuando se disponga de nuevos datos.
